### Paired_t_test (Dependent_test)

## A study has conducted to determine the effectiveness of a weight base program. The table below shows the before and after weights of 10 subject in the program. A) Is the program effective for reducing weight? B) (use a 5% significant level) construct a 95% confidence interval and determine the margin of error.

```python
S.	Bef	Aft
1	185	169
2	192	187
3	206	193
4	177	176
5	225	194
6	168	171
7	256	228
8	239	217
9	199	204
10	218	195
```

```python
## H0-- Program is not effective
## Ha-- Program is effective
```

```python
Before=[185,192,206,177,225,168,256,239,199,218]
After=[169,187,193,176,194,171,228,217,204,195]
```

```python
## tstats= mean_diff/(std_diff/sqrt(n))
```

```python
diff=[]
for i in range(len(After)):
    diff.append(After[i]-Before[i])
diff
```

```python
import numpy as np
import scipy.stats as stats
```

```python
mean= np.mean(diff)
mean
```

```python
std= np.std(diff,ddof=1)
std
```

```python
tstats= mean/(std/np.sqrt(10))
tstats
```

```python
stats.ttest_rel(After,Before)
```

```python
t_critical= stats.t.ppf(0.975,9)
t_critical
```

```python
## pvalue < alpha value and stats value is in the rejection region reject HO
## Program is effective
```

## ANOVA(F-test)

```python
stats.f.ppf(0.95,2,12)
```

```python
a=[2,3,7,2,6]
b=[10,8,7,5,10]
c=[10,13,14,13,15]
```

```python
## Fstats=(Variance_bet_group/df1)/(Variance_within_group/df2)
```

```python
## Variance_bet_group
## sum(overall_mean-mean_samples)**2
df1=2
df2=4+4+4
df2
n=5
```

```python
mean_a= np.mean(a)
mean_b= np.mean(b)
mean_c= np.mean(c)
print('mean_a:',mean_a,', mean_b:',mean_b,', mean_c:',mean_c)
```

```python
Overall_mean= (mean_a+mean_b+mean_c)/3
Overall_mean
```

```python
Variance_bet_group= n*((Overall_mean-mean_a)**2)+n*((Overall_mean-mean_b)**2)+n*((Overall_mean-mean_c)**2)
Variance_bet_group
```

```python
var_within_group= sum((a-mean_a)**2)+ sum((b-mean_b)**2)+sum((c-mean_c)**2)
var_within_group
```

```python
Fstats= (Variance_bet_group/df1)/(var_within_group/df2)
Fstats
```

```python
stats.f_oneway(a,b,c)
```

## A school principal would like to know which day of the week students are most likely to be absent. The principal expects that students will be absent equally during the 5-day school weak. The principal selects a random sample of 100 teachers asking them which day of the week they had the highest number of student absences. The observed and expected results are shown in table below. Based on the results do the days for the highest number of absences occur with equal frequencies? (Use 	a 5% significant level)

	        Mo	Tu	We	Thu Fr
observed	23	16	14	19	28
Expected	20	20	20	20	20


```python
## H0--Absentees are not same on weekdays
## Ha-- Absentess are same on weekdays
## alpha=-0.05, CI=0.95
## n=5
## df=n-1=4
```

```python
chi_critical= stats.chi2.ppf(0.95,4)
chi_critical
```

```python
stats.chisquare(observed,Expected)
```

```python
## Chi_critical>Chi_stats--fail to reject H0, Absentees not same on weekdays
```
