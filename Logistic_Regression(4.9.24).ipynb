```python
target=df[['ADMIT']]
features= df.drop(columns='ADMIT')
```

```python
x_train,x_test,y_train,y_test= train_test_split(features,target,test_size=0.25,random_state=100)
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
log_model= LogisticRegression()
```

```python
log_model.fit(x_train,y_train)
```

```python
y_pred= log_model.predict(x_test)
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
y_pred
```

```python
log_model.predict_proba(x_test)
```

```python
y_pred_new=[1 if x[1]>=0.4 else 0 for x in log_model.predict_proba(x_test)]
```

```python
y_pred_new=[0 if x[0]>=0.6 else 1 for x in log_model.predict_proba(x_test)]
```

```python
accuracy_score(y_test,y_pred_new)
```
